```python
from stemOrchestrator.process import HAADF_tiff_to_png
```

```python
from pathlib import Path

def process_folder(folder_path):
    """
    Process all TIFF files in a folder using your HAADF_tiff_to_png function.
    
    Args:
        folder_path: Path to folder containing TIFF files
    """
    folder = Path(folder_path)
    
    # Check if folder exists
    if not folder.exists():
        print(f"Error: Folder '{folder_path}' does not exist")
        return
    
    # Find all TIFF files
    tiff_files = list(folder.glob("*.tif")) + list(folder.glob("*.tiff"))
    
    if not tiff_files:
        print(f"No TIFF files found in '{folder_path}'")
        return
    
    print(f"Found {len(tiff_files)} TIFF file(s)")
    
    # Process each TIFF file
    for tiff_path in tiff_files:
        try:
            print(f"Processing: {tiff_path.name}")
            HAADF_tiff_to_png(str(tiff_path))
            print(f"✓ Completed: {tiff_path.name}")
        except Exception as e:
            print(f"✗ Error processing {tiff_path.name}: {str(e)}")

# Usage
folder_path = "."  # Replace with your folder path
process_folder(folder_path)
```
